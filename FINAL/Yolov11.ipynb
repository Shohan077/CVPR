```python
import os
import glob
import pandas as pd
import matplotlib.pyplot as plt

def load_results(run_dir):
    csv_file = os.path.join(run_dir, "results.csv")
    if not os.path.exists(csv_file):
        return None
    return pd.read_csv(csv_file)

def plot_metrics(df, run_name, save_dir):
    # Accuracy (Top-1 and Top-5)
    plt.figure()
    plt.plot(df["epoch"], df["metrics/accuracy_top1"], label="Top-1 Accuracy")
    plt.plot(df["epoch"], df["metrics/accuracy_top5"], label="Top-5 Accuracy")
    plt.xlabel("Epochs")
    plt.ylabel("Accuracy")
    plt.title(f"Accuracy vs Epochs ({run_name})")
    plt.legend()
    plt.grid()
    plt.savefig(os.path.join(save_dir, f"{run_name}_accuracy.png"))
    plt.close()

    # Train vs Val loss
    plt.figure()
    plt.plot(df["epoch"], df["train/loss"], label="Train Loss")
    plt.plot(df["epoch"], df["val/loss"], label="Val Loss")
    plt.xlabel("Epochs")
    plt.ylabel("Loss")
    plt.title(f"Loss vs Epochs ({run_name})")
    plt.legend()
    plt.grid()
    plt.savefig(os.path.join(save_dir, f"{run_name}_loss.png"))
    plt.close()

def summarize_last_epoch(df):
    last = df.iloc[-1]
    return {
        "Epoch": int(last["epoch"]),
        "Top-1 Accuracy": round(last["metrics/accuracy_top1"]*100, 2),
        "Top-5 Accuracy": round(last["metrics/accuracy_top5"]*100, 2),
        "Train Loss": round(last["train/loss"], 4),
        "Val Loss": round(last["val/loss"], 4)
    }

def main(runs_root="/content/runs/classify", save_dir="/content/results_summary"):
    os.makedirs(save_dir, exist_ok=True)
    all_runs = sorted(glob.glob(os.path.join(runs_root, "train*")))
    summary = []

    for run in all_runs:
        run_name = os.path.basename(run)
        df = load_results(run)
        if df is None:
            print(f"⚠️ No results.csv found in {run}")
            continue

        # Plot graphs
        plot_metrics(df, run_name, save_dir)

        # Collect last epoch metrics
        metrics = summarize_last_epoch(df)
        metrics["Run"] = run_name
        summary.append(metrics)

    # Save summary table
    if summary:
        summary_df = pd.DataFrame(summary)
        csv_path = os.path.join(save_dir, "summary_table.csv")
        summary_df.to_csv(csv_path, index=False)
        print(f"✅ Summary table saved at: {csv_path}")
        print(summary_df)
    else:
        print("⚠️ No valid training results found!")

if __name__ == "__main__":
    main("/content/runs/classify", "/content/results_summary")
```

```python
# -----------------------------
# 1. Import Libraries
# -----------------------------
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
from torchvision import datasets, transforms, models
import matplotlib.pyplot as plt

# -----------------------------
# 2. Device setup
# -----------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# -----------------------------
# 3. Data Preprocessing
# -----------------------------
# ImageNet mean & std normalization (EfficientNet pretrained expects this)
transform = {
    "train": transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.RandomHorizontalFlip(),
        transforms.RandomRotation(10),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406],
                             [0.229, 0.224, 0.225])
    ]),
    "test": transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406],
                             [0.229, 0.224, 0.225])
    ]),
}

train_dir = "/content/drive/MyDrive/ECG_DATA/train"
test_dir = "/content/drive/MyDrive/ECG_DATA/test"

train_data = datasets.ImageFolder(train_dir, transform=transform["train"])
test_data = datasets.ImageFolder(test_dir, transform=transform["test"])

train_loader = DataLoader(train_data, batch_size=32, shuffle=True)
test_loader = DataLoader(test_data, batch_size=32, shuffle=False)

class_names = train_data.classes
print("Classes:", class_names)

# -----------------------------
# 4. Load Pretrained EfficientNet-B0
# -----------------------------
model = models.efficientnet_b0(pretrained=True)

# Replace classifier head with your dataset classes
num_ftrs = model.classifier[1].in_features
model.classifier[1] = nn.Linear(num_ftrs, len(class_names))

model = model.to(device)

# -----------------------------
# 5. Loss & Optimizer
# -----------------------------
criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(model.parameters(), lr=0.001, momentum=0.9)

# -----------------------------
# 6. Training Function
# -----------------------------
def train_model(model, criterion, optimizer, train_loader, test_loader, epochs=10):
    train_acc, val_acc = [], []

    for epoch in range(epochs):
        model.train()
        correct, total, running_loss = 0, 0, 0.0

        for inputs, labels in train_loader:
            inputs, labels = inputs.to(device), labels.to(device)

            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()

            _, preds = torch.max(outputs, 1)
            correct += (preds == labels).sum().item()
            total += labels.size(0)
            running_loss += loss.item()

        epoch_acc = 100 * correct / total
        train_acc.append(epoch_acc)

        # -----------------------------
        # Validation
        # -----------------------------
        model.eval()
        val_correct, val_total = 0, 0
        with torch.no_grad():
            for inputs, labels in test_loader:
                inputs, labels = inputs.to(device), labels.to(device)
                outputs = model(inputs)
                _, preds = torch.max(outputs, 1)
                val_correct += (preds == labels).sum().item()
                val_total += labels.size(0)

        val_accuracy = 100 * val_correct / val_total
        val_acc.append(val_accuracy)

        print(f"Epoch [{epoch+1}/{epochs}] "
              f"Train Acc: {epoch_acc:.2f}% | Val Acc: {val_accuracy:.2f}% | Loss: {running_loss/len(train_loader):.4f}")

    return train_acc, val_acc

# -----------------------------
# 7. Run Training
# -----------------------------
train_acc, val_acc = train_model(model, criterion, optimizer, train_loader, test_loader, epochs=5)

# -----------------------------
# 8. Plot Accuracy
# -----------------------------
plt.plot(train_acc, label="Train Accuracy")
plt.plot(val_acc, label="Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy (%)")
plt.legend()
plt.grid(True)
plt.show()
```
